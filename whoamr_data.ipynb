```python
country_counts = (
    amr.groupby(["WHORegionName", "AntibioticName"])
    .size()
    .reset_index(name="CountryObservations")
)

country_counts
```

```python
country_amr = (
    amr.groupby(
        ["WHORegionName", "AntibioticName", "CountryTerritoryArea"]
    ) [["InterpretableAST", "Resistant"]]
    .sum()
    .reset_index()
    )

country_amr["ResistancePercentage"] = (
    country_amr["Resistant"]/ country_amr["InterpretableAST"]*100
)

country_amr.head(10)
```

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

model = smf.glm(
    "Resistant + I(InterpretableAST - Resistant) ~ C(WHORegionName) + C(AntibioticName)" ,
    data=country_amr,
    family=sm.families.Binomial()
).fit()
print(model.summary())

```

```python
dispersion = model.pearson_chi2 / model.df_resid

print("Pearson chi-square:", model.pearson_chi2)
print("Residual degrees of freedom:", model.df_resid)
print("Dispersion:", dispersion)
```

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

robust_model = smf.glm(
    "Resistant + I(InterpretableAST - Resistant) ~ C(WHORegionName) + C(AntibioticName)",
    data=country_amr,
    family=sm.families.Binomial()
).fit(
    cov_type="cluster",
    cov_kwds={"groups":
              country_amr["CountryTerritoryArea"]}
)

print(robust_model.summary())
```

```python
import numpy as np
import pandas as pd

results = pd.DataFrame({
    "Coefficient": robust_model.params,
    "Odds_Ratio": np.exp(robust_model.params),
    "CI_Lower": np.exp(robust_model.conf_int()[0]),
    "CI_Upper": np.exp(robust_model.conf_int()[1]),
    "P_value": robust_model.pvalues


})

results
```

```python
import matplotlib.pyplot as plt

# The error is because 'clean_results' is not defined. It should be 'results'.
# Assuming the user intends to plot region and antibiotic effects,
# the 'Intercept' row should be dropped from the 'results' DataFrame.
plot_data_original_order = results.drop('Intercept').copy()

# Generate labels dynamically from the index of the plot_data DataFrame
# to ensure they match the data being plotted.
# Clean up the index names for better plot labels
labels_original_order = [
    item.replace('C(WHORegionName)[T.', '').replace(']', '')
    .replace('C(AntibioticName)[T.', '')
    .replace('Eastern Mediterranean Region', 'Eastern Mediterranean')
    .replace('European Region', 'European')
    .replace('Region of the Americas', 'Americas')
    .replace('South-East Asia Region', 'South-East Asia')
    .replace('Western Pacific Region', 'Western Pacific')
    for item in plot_data_original_order.index
]

# For plotting a forest plot, it's common to reverse the order so the
# first category appears at the top of the plot.
plot_data = plot_data_original_order.iloc[::-1].copy()
labels = labels_original_order[::-1]

plt.figure(figsize=(9,7))

plt.errorbar(
    plot_data["Odds_Ratio"],
    range(len(plot_data)),
    xerr=[
        plot_data["Odds_Ratio"] - plot_data["CI_Lower"],
        plot_data["CI_Upper"] - plot_data["Odds_Ratio"]
    ],
    fmt="o",
    capsize=4
)

plt.axvline(1, linestyle="--")
plt.yticks(range(len(plot_data)), labels)
plt.xscale("log")
plt.xlabel("Adjusted odds ratio (log scale)")
plt.ylabel("")
plt.title("Adjusted odds of antimicrobial resistance")
plt.tight_layout()
plt.show()
```
